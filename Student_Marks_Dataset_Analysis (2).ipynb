# Student Marks Dataset Analysis

## Week 1 – Python for Data Science Basics

### Objective

The objective of this project is to analyze a Student Marks dataset using Python, Pandas, and NumPy. The project focuses on basic data exploration, checking and handling missing data, and calculating the average, maximum, and minimum scores for Mathematics, Reading, and Writing.


Dataset Description

The dataset contains performance information for 1,000 students across different demographic and educational categories.

The dataset contains 8 columns:

Gender
Race/Ethnicity
Parental Level of Education
Lunch
Test Preparation Course
Math Score
Reading Score
Writing Score

The three score columns are numerical and are used for the main analysis.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("/StudentsPerformance.csv")
```

```python
df.head()
```

## Data Exploration

The dataset was inspected to understand its structure, number of records, column names, and data types.


```python
df.info()
```

```python
df.isnull().sum()
```

### Missing Data Check

The dataset was checked for missing values using `isnull().sum()`. All columns contain 1,000 non-null values, indicating that there are no missing values in the dataset. Therefore, no rows needed to be removed.


## Score Analysis

The average, maximum, and minimum scores were calculated for Mathematics, Reading, and Writing using Pandas.


```python
results = pd.DataFrame({
    "Average": df[["math score", "reading score", "writing score"]].mean(),
    "Maximum": df[["math score", "reading score", "writing score"]].max(),
    "Minimum": df[["math score", "reading score", "writing score"]].min()
})
results
```

## NumPy Analysis

NumPy was used to calculate the overall average score across the three subjects. The three score columns were converted into a NumPy array for numerical analysis.


```python
scores = df[["math score", "reading score", "writing score"]].to_numpy()

overall_average = np.mean(scores)

overall_average
```

## Findings and Insights

The analysis shows that the Reading score had the highest average at 69.169, followed by Writing at 68.054 and Mathematics at 66.089. The maximum score in all three subjects was 100.

The minimum scores were 0 in Mathematics, 17 in Reading, and 10 in Writing. The overall average across the three subjects was approximately 67.77.

The dataset contained no missing values, so no rows needed to be removed during the data-cleaning process.
