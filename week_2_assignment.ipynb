# Exploratory Data Analysis of COVID-19 Dataset

## Week 2 – Data Visualization

### Objective
To perform exploratory data analysis on COVID-19 data using Pandas, Matplotlib, and Seaborn.
The analysis includes data cleaning, identifying the top 5 countries by confirmed cases,
correlation analysis, and different data visualizations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('Downloads/country_wise_latest (1).csv')
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
df.isnull().sum()
```

## Data Cleaning

The dataset was checked for missing values using `isnull().sum()`.
No missing values were found in any of the 15 columns, so no missing-value treatment was required.

```python
top_5=df.nlargest(5,'Confirmed')
top_5[['Country/Region','Confirmed']]
```

```python
plt.figure(figsize=(8,5))
plt.bar(top_5['Country/Region'], top_5['Confirmed'])
plt.title('Top 5 Countries by Confirmed COVID-19 Cases')
plt.xlabel('Country')
plt.ylabel('Confirmed')
plt.show()
```

```python
correlation = df.corr(numeric_only=True)
plt.figure(figsize=(12, 8))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap of COVID-19 Data')
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df['Confirmed'], df['Deaths'])

plt.title('Confirmed COVID-19 Cases vs Deaths')
plt.xlabel('Confirmed Cases')
plt.ylabel('Deaths')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df['Confirmed'], bins=20)

plt.title('Distribution of Confirmed COVID-19 Cases')
plt.xlabel('Confirmed Cases')
plt.ylabel('Number of Countries')

plt.show()
```

## Key Findings

1. The United States had the highest number of confirmed COVID-19 cases among the top 5 countries in the dataset.
2. Brazil, India, Russia, and South Africa followed the United States in confirmed cases.
3. The correlation heatmap shows the relationships between different numerical COVID-19 variables.
4. The scatter plot shows the relationship between confirmed cases and deaths across countries.
5. The histogram shows the distribution of confirmed COVID-19 cases among the countries.
6. No missing values were found in the dataset.
